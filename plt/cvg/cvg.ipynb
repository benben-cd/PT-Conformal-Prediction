```python
## group coverage result plot
import matplotlib.pyplot as plt
import pandas as pd
import os

print(os.getcwd())
root_path = 'd:/学习材料/科研项目/Sparse_CP'
os.chdir(root_path)


##'meps_19', 'meps_20', 'meps_21', 'blog_data', 'bio', 'facebook_1', 'facebook_2', 'concrete', 'star'
dataset_names = ['bike']
alpha_list = [0.05, 0.1, 0.15, 0.2]
p_list = [0.96, 0.97, 0.98, 0.99, 1]

for dataset_name in dataset_names:
    if dataset_name == 'bike':
        group_names = ['day', 'month', 'year']
    for group_name in group_names:
        file_path = f'./ordinary_regression_task/group_coverage_result/gc_{dataset_name}_{group_name}.csv'
        ablation_p_reg = pd.read_csv(file_path)
        data_list = [ablation_p_reg]

        linestyle_list = ['--', '-']
        color_list = ['orange', 'red', 'blue', 'purple']
        fontsize = 16

        # 分别保存四张图
        for i in range(len(data_list)):
            # 创建一个新的画布，每次绘制一个单独的图
            fig, ax = plt.subplots(figsize=(8, 6))  # 修改为单独的图，大小适中
            df = data_list[i]
            
            # 根据不同的子图选择x轴
            if i == 0:
                x_axis = p_list
                ax.set_xlabel('p', fontsize=fontsize)
                ax.set_ylabel('coverage', fontsize=fontsize)
            
            # 设置画布背景和网格
            ax.set_facecolor('whitesmoke')
            ax.grid(True, color='lightgrey')
            ax.tick_params(axis='both', labelsize=fontsize)
            # 绘制每个折线图
            for j in range(len(alpha_list)):
                color = color_list[j]
                ax.plot(df.iloc[j], marker='o', linestyle=linestyle_list[0], color=color, label=f'VCP_$\\alpha={alpha_list[j]}$')
                ax.plot(df.iloc[j + 4], marker='o', linestyle=linestyle_list[1], color=color, label=f'PT_$\\alpha={alpha_list[j]}$')
            
            # 添加图例
            ax.legend(loc='upper right', fontsize=14)
            
            # 保存每张图为独立文件
            plt.savefig(f'plt/cvg/gc_{dataset_name}_{group_name}.pdf')  # 保存每个图到独立文件
            plt.close(fig)  # 关闭当前图形，避免所有图形显示在同一个画布上
```
